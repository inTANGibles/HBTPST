# Benchmark: train five methods and plot metrics

- Run from **project root** (folder that contains `grid_world/`).
- Methods: `bc`, `env_lstm`, `maxent_irl`, `gail`, `medirl` (same `paths` + `grid` -> unified **f(s)** via `features_arr`).
- Reduce `MEDIRL_QUICK_EPOCHS` in the train cell for smoke tests only; paper runs use full `n_epochs` from yaml.
- All methods share **70/15/15** traj split (`split` in yaml) and report metrics on the **test** partition.

```python
from __future__ import annotations

import json
import sys
import traceback
from copy import deepcopy
from datetime import datetime
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

_here = Path.cwd().resolve()
if (_here / "grid_world").is_dir():
    ROOT = _here
elif (_here.parent / "grid_world").is_dir():
    ROOT = _here.parent
else:
    raise RuntimeError("Set cwd to project root (contains grid_world).")

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
print("ROOT =", ROOT)
```

```python
RUN_TAG = datetime.now().strftime("%Y%m%d_%H%M%S")
BATCH_DIR = ROOT / "benchmark" / "outputs" / f"batch_{RUN_TAG}"
BATCH_DIR.mkdir(parents=True, exist_ok=True)

CONFIGS = {
    "bc": ROOT / "benchmark" / "configs" / "bc.yaml",
    "env_lstm": ROOT / "benchmark" / "configs" / "env_lstm.yaml",
    "maxent_irl": ROOT / "benchmark" / "configs" / "maxent_irl.yaml",
    "gail": ROOT / "benchmark" / "configs" / "gail.yaml",
    "medirl": ROOT / "benchmark" / "configs" / "medirl.yaml",
}

MEDIRL_QUICK_EPOCHS = None  # set e.g. 20 for smoke test; None = use yaml n_epochs (500)
METHOD_ORDER = ["bc", "env_lstm", "maxent_irl", "gail", "medirl"]
print("BATCH_DIR:", BATCH_DIR)
```

```python
from benchmark.common.utils import load_config, save_json, set_seed
from benchmark.methods import get_train_eval


def run_one(method: str, config_path: Path, out_dir: Path) -> dict:
    out_dir.mkdir(parents=True, exist_ok=True)
    cfg = load_config(config_path)
    seed = int(cfg.get("seed", 110))
    set_seed(seed)
    if method == "medirl" and MEDIRL_QUICK_EPOCHS is not None:
        cfg = deepcopy(cfg)
        cfg.setdefault("medirl", {})["n_epochs"] = int(MEDIRL_QUICK_EPOCHS)
    cfg["_output_dir"] = str(out_dir)
    import shutil

    shutil.copy2(config_path, out_dir / "config_snapshot.yaml")
    save_json(out_dir / "run_meta.json", {"method": method, "seed": seed, "config_path": str(config_path)})
    try:
        train_fn, _, _ = get_train_eval(method)
        result = train_fn(cfg, output_dir=out_dir)
        metrics = result.get("metrics", {})
        save_json(out_dir / "train_result.json", {k: v for k, v in result.items() if k != "metrics"})
        return {"status": "ok", "metrics": metrics, "error": None, "output_dir": str(out_dir)}
    except NotImplementedError as e:
        return {"status": "skipped", "metrics": {}, "error": str(e), "output_dir": str(out_dir)}
    except Exception:
        return {"status": "error", "metrics": {}, "error": traceback.format_exc(), "output_dir": str(out_dir)}


results = []
for m in METHOD_ORDER:
    cfg_path = CONFIGS[m]
    if not cfg_path.is_file():
        results.append({"method": m, "status": "missing_config", "metrics": {}, "error": str(cfg_path), "output_dir": ""})
        print(m, "SKIP: no config")
        continue
    out_m = BATCH_DIR / m
    print("===", m, "===>", out_m)
    r = run_one(m, cfg_path, out_m)
    r["method"] = m
    results.append(r)
    print("   status:", r["status"])
    if r["error"] and r["status"] != "skipped":
        print(str(r["error"])[:800])

save_json(BATCH_DIR / "batch_summary.json", results)
print("Saved", BATCH_DIR / "batch_summary.json")
```

```python
import json
from pathlib import Path

from benchmark.common import paper_metrics as _pm

PAPER_LABELS = _pm.PAPER_LABELS
PAPER_METRIC_ORDER = _pm.PAPER_METRIC_ORDER
# 与 benchmark.common.paper_metrics.PAPER_TABLE_FORCE_COLUMNS 保持一致；旧内核无该名时用默认值
PAPER_TABLE_FORCE_COLUMNS = getattr(_pm, "PAPER_TABLE_FORCE_COLUMNS", ("R2", "RMSE", "EVD"))

rows = []
for r in results:
    met = dict(r.get("metrics") or {})
    # 与内存中的 results 合并：优先读各方法目录下最新的 metrics.json（只重跑本格或改过 paper_metrics 后仍可对齐磁盘）
    od = r.get("output_dir") or ""
    if od:
        mp = Path(od) / "metrics.json"
        if mp.is_file():
            try:
                met.update(json.loads(mp.read_text(encoding="utf-8")))
            except (OSError, json.JSONDecodeError):
                pass
    row = {"method": r["method"], "status": r.get("status")}
    for k in PAPER_METRIC_ORDER:
        row[k] = met.get(k)
    rows.append(row)

df_raw = pd.DataFrame(rows).set_index("method").reindex(METHOD_ORDER)
present = [k for k in PAPER_METRIC_ORDER if k in PAPER_TABLE_FORCE_COLUMNS or df_raw[k].notna().any()]
df = df_raw[["status"] + present]
display_df = df.rename(columns={k: PAPER_LABELS[k] for k in present})
display(display_df)
display_df.to_csv(BATCH_DIR / "metrics_table.csv")
```

```python
from benchmark.common.paper_metrics import PAPER_LABELS, PAPER_METRIC_ORDER

plot_df = df.reset_index()
ok_mask = plot_df["status"] == "ok"
keys = [k for k in PAPER_METRIC_ORDER if k in df.columns and k != "status" and df[k].notna().any()]
n = len(keys)
if n == 0:
    print("No paper metric columns to plot (re-run training and table cell first).")
else:
    fig, axes = plt.subplots(1, n, figsize=(max(10, 3 * n), 4), squeeze=False)
    axes = axes.ravel()
    for ax, key in zip(axes, keys):
        sub = plot_df.loc[ok_mask & plot_df[key].notna()]
        if len(sub):
            ax.bar(sub["method"], sub[key], color="steelblue")
        else:
            ax.text(0.5, 0.5, "no data", ha="center", va="center", transform=ax.transAxes)
        ax.set_title(PAPER_LABELS[key])
        ax.tick_params(axis="x", rotation=25)
    for j in range(len(keys), len(axes)):
        axes[j].set_visible(False)
    plt.suptitle(f"batch {RUN_TAG}", y=1.02)
    plt.tight_layout()
    fig_path = BATCH_DIR / "metrics_comparison.png"
    plt.savefig(fig_path, dpi=150, bbox_inches="tight")
    plt.show()
    print("Figure:", fig_path)
```

- Table uses paper metrics in fixed order (**R²↑ (SVF)**, **RMSE↓ (SVF)**, **SVF-KL↓**, **EMD↓**, **EVD↓ (SVF MAE)**, **Reward-Corr↑**, **Feat. Match Err↓**). **R² / RMSE / EVD** compare expert empirical SVF to **expected SVF** under the method's policy (same chain as ``DMEIRL.Expected_StateVisitationFrequency``): reward methods use VI softmax from learned reward; BC / GAIL / Env-LSTM use saved policy probabilities. **Reward-Corr** still requires `real_reward_arr` plus learned reward. **SVF-KL / EMD / Feat. Match** use rollout **empirical** SVF vs expert. Bars still skip metrics that have no numeric value for any successful run.
- **gail** / **env_lstm** use the same `features_arr` as **bc** when configs share `paths` + `grid`.
